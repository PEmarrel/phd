```python
word_to_test = word_GNG_IWW_inter_gg
res = np.zeros((len(word_to_test), len(word_to_test)), dtype=float)


for i, w1 in enumerate(word_to_test):
    for j, w2 in enumerate(word_to_test):
        similarity_score = model.similarity(w1, w2)
        res[i][j] = similarity_score * 100 // 1

res *= 0.01
df = pd.DataFrame(res, columns=word_to_test, index=word_to_test)
print(df)

# 1. Convert Similarity to Distance (Distance = 1 - Similarity)
# We make sure the diagonal is strictly 0.0 to avoid math errors
dist_matrix = 1.0 - df.values
np.fill_diagonal(dist_matrix, 0.0)

# 2. Convert to a "condensed" distance matrix (required by scipy)
condensed_dist = ssd.squareform(dist_matrix)

# 3. Perform Hierarchical Clustering
# 'complete' linkage means it prioritizes clusters where ALL members are close to each other
Z = linkage(condensed_dist, method='complete')

# 4. Extract clusters based on a distance threshold
# For example, threshold = 0.3 means we want words with at least 0.70 similarity (1 - 0.3)
distance_threshold = 0.7
cluster_labels = fcluster(Z, t=distance_threshold, criterion='distance')

# 5. Group the words by their cluster label to see the results
clusters = {}
for word, label in zip(word_to_test, cluster_labels):
    if label not in clusters:
        clusters[label] = []
    clusters[label].append(word)

# Find the largest/best cluster
best_cluster_id = max(clusters, key=lambda k: len(clusters[k]))
print(f"The largest tight cluster (minimum similarity 0.50) contains:\n{clusters[best_cluster_id]}")

cluster_cohesion = {}

for cluster_id, words in clusters.items():
    # We only care about clusters with 3 or more words
    if len(words) >= 5:
        # 1. Extract the sub-matrix from your DataFrame just for these words
        sub_matrix = df.loc[words, words]
        
        # 2. Get the upper triangle to average the pairwise similarities (ignoring the 1.0 diagonal)
        mask = np.triu(np.ones(sub_matrix.shape), k=1).astype(bool)
        avg_similarity = sub_matrix.values[mask].mean()
        
        cluster_cohesion[cluster_id] = avg_similarity

# Sort clusters by their average similarity in descending order
sorted_by_cohesion = sorted(cluster_cohesion.items(), key=lambda x: x[1], reverse=True)

print("\n--- Top Clusters by Cohesion (Tightness) ---")
# Print the top 5 tightest clusters
for cluster_id, avg_sim in sorted_by_cohesion[:]:
    words = clusters[cluster_id]
    print(f"Cluster {cluster_id} (Avg Sim: {avg_sim:.3f}, Size: {len(words)}): {words}")
```

```python
word_to_test = word_GNG_IWW_inter_gg
res = np.zeros((len(word_to_test), len(word_to_test)), dtype=float)


for i, w1 in enumerate(word_to_test):
    for j, w2 in enumerate(word_to_test):
        similarity_score = model.similarity(w1, w2)
        res[i][j] = similarity_score

df = pd.DataFrame(res, columns=word_to_test, index=word_to_test)

# 1. Convert Similarity to Distance (Distance = 1 - Similarity)
# We make sure the diagonal is strictly 0.0 to avoid math errors
dist_matrix = 1.0 - df.values
np.fill_diagonal(dist_matrix, 0.0)

# 2. Convert to a "condensed" distance matrix (required by scipy)
condensed_dist = ssd.squareform(dist_matrix)

# 3. Perform Hierarchical Clustering
# 'complete' linkage means it prioritizes clusters where ALL members are close to each other
Z = linkage(condensed_dist, method='complete')

# 4. Extract clusters based on a distance threshold
# For example, threshold = 0.3 means we want words with at least 0.70 similarity (1 - 0.3)
distance_threshold = 0.7
cluster_labels = fcluster(Z, t=distance_threshold, criterion='distance')

# 5. Group the words by their cluster label to see the results
clusters = {}
for word, label in zip(word_to_test, cluster_labels):
    if label not in clusters:
        clusters[label] = []
    clusters[label].append(word.tolist())

# Find the largest/best cluster
best_cluster_id = max(clusters, key=lambda k: len(clusters[k]))
print(f"The largest tight cluster (minimum similarity 0.50) contains:\n{clusters[best_cluster_id]}")

cluster_cohesion = {}

for cluster_id, words in clusters.items():
    # We only care about clusters with 3 or more words
    if len(words) >= 5:
        # 1. Extract the sub-matrix from your DataFrame just for these words
        sub_matrix = df.loc[words, words]
        
        # 2. Get the upper triangle to average the pairwise similarities (ignoring the 1.0 diagonal)
        mask = np.triu(np.ones(sub_matrix.shape), k=1).astype(bool)
        avg_similarity = sub_matrix.values[mask].mean()
        
        cluster_cohesion[cluster_id] = avg_similarity

# Sort clusters by their average similarity in descending order
sorted_by_cohesion = sorted(cluster_cohesion.items(), key=lambda x: x[1], reverse=True)

print("\n--- Top Clusters by Cohesion (Tightness) ---")
# Print the top 5 tightest clusters
for cluster_id, avg_sim in sorted_by_cohesion[:]:
    words = clusters[cluster_id]
    print(f"Cluster {cluster_id} (Avg Sim: {avg_sim:.3f}, Size: {len(words)}): {words}")
```
